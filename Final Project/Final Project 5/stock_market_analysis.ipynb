```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Stock_Market.csv')
df
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.duplicated().sum()
```

```python
plt.figure(figsize=(10,5))
plt.plot(df["Close"])
plt.title("Closing Price Trend")
plt.xlabel("Index")
plt.ylabel("Close Price")
plt.show()
```

```python
df["Volume"].head(20).plot(kind="bar")
plt.title("Top 20 Trading Volume")
plt.xlabel("Index")
plt.ylabel("Volume")
plt.show()
```

```python
plt.hist(df["Close"], bins=15)
plt.title("Closing Price Distribution")
plt.xlabel("Close Price")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(y=df["Close"])
plt.title("Closing Price Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,6))
corr = df[["Open","High","Low","Close","Volume","Dividends"]].corr()
sns.heatmap(corr,annot=True,fmt="2f",cmap="coolwarm",linewidths=0.5,square=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(df["Open"].head(300), df["Close"].head(300),alpha=0.7)
plt.title("Open Price vs Close Price")
plt.xlabel("Open Price")
plt.ylabel("Close Price")
plt.show()
```
