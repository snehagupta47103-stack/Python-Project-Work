```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("Covid19.csv")
df.head()
```

```python
id_cols = ['Province/States', 'Country/Region', 'WHO region', 'WHO region label']
date_cols = [c for c in df.columns if c not in id_cols]

df_long = df.melt(id_vars=id_cols, value_vars=date_cols, var_name='Date', value_name='Confirmed')

df_long['Date'] = pd.to_datetime(df_long['Date'], format='mixed')

df_long['Confirmed'] = df_long['Confirmed'].fillna(0)
df_long.drop_duplicates(inplace=True)

df_long.head()
```

```python
df_long[['Confirmed']].describe()
```

```python
global_cases = df_long.groupby('Date')['Confirmed'].sum()
global_cases
```

```python
plt.figure(figsize=(10, 5))
global_cases.plot(kind='line', color='red')
plt.title('Global COVID-19 Confirmed Cases Over Time')
plt.xlabel('Date')
plt.ylabel('Confirmed Cases')
plt.grid(True)
plt.show()
```

```python
valid_dates = df_long.groupby('Date')['Confirmed'].sum()
latest_valid_date = valid_dates[valid_dates > 0].index.max()

top_countries = df_long[df_long['Date'] == latest_valid_date].groupby('Country/Region')['Confirmed'].sum().nlargest(5)
top_countries
```

```python
plt.figure(figsize=(10, 5))
top_countries.plot(kind='bar', color='purple')
plt.title(f'Top 5 Most Affected Countries ({latest_valid_date.strftime("%Y-%m-%d")})')
plt.xlabel('Country/Region')
plt.ylabel('Confirmed Cases')
plt.xticks(rotation=45)
plt.show()
```

```python
region_cases = df_long[df_long['Date'] == latest_valid_date].groupby('WHO region')['Confirmed'].sum()
region_cases
```

```python
plt.figure(figsize=(10, 5))
region_cases.plot(kind='bar', color='orange')
plt.title('Total Confirmed Cases per WHO Region')
plt.xlabel('WHO Region')
plt.ylabel('Confirmed Cases')
plt.show()

growth_region = region_cases.idxmax()
growth_value = region_cases.max()
print(f'The WHO region with highest confirmed cases is {growth_region} with total cases of {int(growth_value)}.')
```
