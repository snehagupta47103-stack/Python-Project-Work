```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('Air_Quality.csv')
df
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df.describe()
```

```python
df.duplicated().sum()
```

```python
df.replace(-200, np.nan, inplace=True)
```

```python
plt.figure(figsize=(10,5))
plt.plot(df["CO(GT)"].head(200))
plt.title("CO(GT) Analysis")
plt.xlabel("Index")
plt.ylabel("CO(GT)")
plt.show()
```

```python
df["NO2(GT)"].head(20).plot(kind="bar")
plt.title("NO2(GT) Analysis")
plt.xlabel("Index")
plt.ylabel("NO2(GT)")
plt.show()
```

```python
plt.hist(df["T"].dropna(), bins=15)
plt.title("Temperature Distribution")
plt.xlabel("Temperature")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(12,4))
sns.heatmap(df.corr(numeric_only=True),annot=True,cmap="Blues")
plt.title("Correlation Heatmap")
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.scatter(df["T"].head(300), df["CO(GT)"].head(300))
plt.title("Temperature vs CO(GT)")
plt.xlabel("Temperature")
plt.ylabel("CO(GT)")
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.boxplot(y=df["NO2(GT)"])
plt.title("NO2(GT) Distribution")
plt.show()
```
