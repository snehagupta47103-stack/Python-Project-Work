```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


class LibraryDashboard:

    def __init__(self):
        self.data = None

    def load_data(self, file_path):
        try:
            self.data = pd.read_csv(file_path)

            required_columns = [
                "Transaction ID",
                "Date",
                "User ID",
                "Book Title",
                "Genre",
                "Borrowing Duration (Days)"
            ]

            for col in required_columns:
                if col not in self.data.columns:
                    raise Exception(f"Missing Column: {col}")

            self.data.drop_duplicates(inplace=True)
            self.data.dropna(inplace=True)

            self.data["Date"] = pd.to_datetime(
                self.data["Date"],
                format="%d-%m-%Y"
            )

            print("\nDataset Loaded Successfully")
            print(self.data.head())

        except Exception as e:
            print("Error:", e)

    def calculate_statistics(self):

        avg_duration = np.mean(
            self.data["Borrowing Duration (Days)"]
        )

        std_duration = np.std(
            self.data["Borrowing Duration (Days)"]
        )

        most_book = (
            self.data["Book Title"]
            .value_counts()
            .idxmax()
        )

        busiest_day = (
            self.data["Date"]
            .dt.day_name()
            .value_counts()
            .idxmax()
        )

        print("\n===== Statistics =====")
        print("Average Borrowing Duration:", round(avg_duration, 2))
        print("Standard Deviation:", round(std_duration, 2))
        print("Most Borrowed Book:", most_book)
        print("Busiest Day:", busiest_day)

    def filter_transactions(self):

        print("\n1. Filter by Genre")
        print("2. Filter by Minimum Borrowing Days")

        choice = input("Enter Choice: ")

        if choice == "1":

            genre = input("Enter Genre: ")

            result = self.data[
                self.data["Genre"].str.lower() == genre.lower()
            ]

            print(result)

        elif choice == "2":

            days = int(input("Enter Minimum Days: "))

            result = self.data[
                self.data["Borrowing Duration (Days)"] >= days
            ]

            print(result)

        else:
            print("Invalid Choice")

    def bar_chart(self):

        plt.figure(figsize=(8,5))

        self.data["Genre"].value_counts().plot(
            kind="bar",
            color="skyblue"
        )

        plt.title("Books Borrowed by Genre")
        plt.xlabel("Genre")
        plt.ylabel("Number of Books")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()


    def line_graph(self):

        daily = self.data.groupby("Date").size()

        plt.figure(figsize=(8,5))

        plt.plot(
            daily.index,
            daily.values,
            marker="o"
        )

        plt.title("Daily Borrowing Trend")
        plt.xlabel("Date")
        plt.ylabel("Transactions")
        plt.xticks(rotation=45)
        plt.grid(True)

        plt.tight_layout()
        plt.show()


    def pie_chart(self):

        genre = self.data["Genre"].value_counts()

        plt.figure(figsize=(6,6))

        plt.pie(
            genre.values,
            labels=genre.index,
            autopct="%1.1f%%",
            startangle=90
        )

        plt.title("Genre Distribution")
        plt.show()


    def heatmap(self):

        table = pd.crosstab(
            self.data["Genre"],
            self.data["Borrowing Duration (Days)"]
        )

        plt.figure(figsize=(10,6))

        sns.heatmap(
            table,
            annot=True,
            cmap="YlGnBu"
        )

        plt.title("Genre vs Borrowing Duration")

        plt.tight_layout()
        plt.show()


    def report(self):

        print("\n========== LIBRARY REPORT ==========")

        print("Total Transactions:",
              len(self.data))

        print("Total Users:",
              self.data["User ID"].nunique())

        print("Total Books:",
              self.data["Book Title"].nunique())

        print("\nTop 5 Borrowed Books")
        print(
            self.data["Book Title"]
            .value_counts()
            .head()
        )

        print("\nGenre Wise Count")
        print(
            self.data["Genre"]
            .value_counts()
        )
def main():

    dashboard = LibraryDashboard()

    dashboard.load_data("library_transactions.csv")

    if dashboard.data is None:
        print("CSV file could not be loaded.")
        return

    while True:

        print("\n========== E-LIBRARY DASHBOARD ==========")
        print("1. Show Statistics")
        print("2. Filter Transactions")
        print("3. Bar Chart")
        print("4. Line Graph")
        print("5. Pie Chart")
        print("6. Heatmap")
        print("7. Report")
        print("8. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            dashboard.calculate_statistics()

        elif choice == "2":
            dashboard.filter_transactions()

        elif choice == "3":
            dashboard.bar_chart()

        elif choice == "4":
            dashboard.line_graph()

        elif choice == "5":
            dashboard.pie_chart()

        elif choice == "6":
            dashboard.heatmap()

        elif choice == "7":
            dashboard.report()

        elif choice == "8":
            print("Thank You!")
            break

        else:
            print("Invalid Choice! Please Try Again.")


if __name__ == "__main__":
    main()
```
