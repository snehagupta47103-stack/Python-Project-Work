```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('sales_data.csv')
df
```

```python
class SalesDataAnalyzer:

    def __init__(self):
        self.data = None
        self.current_plot = None

    def __del__(self):
        print("Program Closed Successfully.")

    def load_data(self):

        file = input("Enter CSV file path : ")

        try:
            self.data = pd.read_csv(file)
            print("\nDataset Loaded Successfully!\n")

        except Exception as e:
            print("Error :", e)

    def explore_data(self):

        while True:

            print("\n========== Explore Data ==========")
            print("1. Display First 5 Rows")
            print("2. Display Last 5 Rows")
            print("3. Display Column Names")
            print("4. Display Data Types")
            print("5. Display Basic Info")
            print("6. Back")

            choice = int(input("Enter your choice : "))

            if choice == 1:
                print(self.data.head())

            elif choice == 2:
                print(self.data.tail())

            elif choice == 3:
                print(self.data.columns)

            elif choice == 4:
                print(self.data.dtypes)

            elif choice == 5:
                self.data.info()

            elif choice == 6:
                break

            else:
                print("Invalid Choice")

    def clean_data(self):

        while True:

            print("\n========== HANDLE MISSING DATA ==========")
            print("1. Display Missing Values")
            print("2. Fill Missing Values with Mean")
            print("3. Drop Missing Values")
            print("4. Replace Missing Values")
            print("5. Back")

            choice = int(input("Enter your choice : "))

            if choice == 1:
                print(self.data.isnull().sum())

            elif choice == 2:
                numeric = self.data.select_dtypes(include=np.number).columns
                self.data[numeric] = self.data[numeric].fillna(
                    self.data[numeric].mean()
                )
                print("\nMissing values filled successfully.")

            elif choice == 3:
                self.data.dropna(inplace=True)
                print("\nMissing values removed successfully.")

            elif choice == 4:
                value = input("Enter replacement value : ")
                self.data.fillna(value, inplace=True)
                print("\nMissing values replaced successfully.")

            elif choice == 5:
                break

            else:
                print("Invalid Choice")

    def mathematical_operations(self):
        print("\n========== MATHEMATICAL OPERATIONS ==========\n")
        print("Sum")
        print(self.data.sum(numeric_only=True))
        print("\nMean")
        print(self.data.mean(numeric_only=True))
        print("\nMaximum")
        print(self.data.max(numeric_only=True))
        print("\nMinimum")
        print(self.data.min(numeric_only=True))

    def combine_data(self):
        file = input("Enter second CSV file : ")

        try:
            df2 = pd.read_csv(file)
            combined = pd.concat([self.data, df2], ignore_index=True)
            print(combined)

        except Exception as e:
            print("Error :", e)

    def split_data(self):
        print("\nFirst Half\n")
        print(self.data.iloc[:len(self.data)//2])
        print("\nSecond Half\n")
        print(self.data.iloc[len(self.data)//2:])

    def search_sort_filter(self):
        while True:
            print("\n========== SEARCH / SORT / FILTER ==========")
            print("1. Search")
            print("2. Sort")
            print("3. Filter")
            print("4. Back")

            choice = int(input("Enter your choice : "))

            if choice == 1:
                column = input("Enter column name : ")
                value = input("Enter value : ")
                print(self.data[self.data[column].astype(str) == value])

            elif choice == 2:
                column = input("Enter column name : ")
                print(self.data.sort_values(by=column))

            elif choice == 3:
                column = input("Enter column name : ")
                value = input("Enter value : ")
                print(self.data[self.data[column].astype(str) == value])

            elif choice == 4:
                break

            else:
                print("Invalid Choice")

    def aggregate_functions(self):
        print("\n========== AGGREGATE FUNCTIONS ==========\n")
        region_sales = self.data.groupby("Region")["Sales"].agg(["sum","mean","count","max","min"])
        print(region_sales)


    def statistical_analysis(self):
        print("\n========== STATISTICAL ANALYSIS ==========\n")
        print(self.data.describe())
        print("\nStandard Deviation\n")
        print(self.data.std(numeric_only=True))
        print("\nVariance\n")
        print(self.data.var(numeric_only=True))
        print("\nMedian\n")
        print(self.data.median(numeric_only=True))
        print("\nQuantiles\n")
        print(self.data.quantile([0.25,0.50,0.75],numeric_only=True))

    def create_pivot_table(self):
        print("\n========== PIVOT TABLE ==========\n")
        pivot = pd.pivot_table(
            self.data,
            values="Sales",
            index="Region",
            columns="Year",
            aggfunc="sum",
            fill_value=0
        )
        print(pivot)

    def numpy_operations(self):
        print("\n========== NUMPY OPERATIONS ==========\n")
        arr = self.data["Sales"].to_numpy()
        print("Array")
        print(arr)
        print("\nFirst Element")
        print(arr[0])
        print("\nFirst Five Elements")
        print(arr[:5])
        print("\nMean :",np.mean(arr))
        print("Maximum :",np.max(arr))
        print("Minimum :",np.min(arr))
        print("Sum :",np.sum(arr))
        print("Standard Deviation :",np.std(arr))

    def visualize_data(self):
        while True:
            print("\n========== DATA VISUALIZATION ==========")
            print("1. Bar Plot")
            print("2. Line Plot")
            print("3. Scatter Plot")
            print("4. Pie Chart")
            print("5. Histogram")
            print("6. Stack Plot")
            print("7. Heatmap")
            print("8. Box Plot")
            print("9. Back")

            choice = int(input("Enter your choice : "))

            if choice == 1:
                plt.figure(figsize=(8,5))
                graph = self.data.groupby("Region")["Sales"].sum()
                plt.bar(graph.index,graph.values)
                plt.title("Sales by Region")
                plt.xlabel("Region")
                plt.ylabel("Sales")
                plt.show()

            elif choice == 2:
                year_sales = self.data.groupby("Year")["Sales"].sum()
                plt.figure(figsize=(8,5))
                plt.plot(year_sales.index,year_sales.value, marker="o")
                plt.title("Year Wise Sales")
                plt.xlabel("Year")
                plt.ylabel("Sales")
                plt.grid(True)
                plt.show()

            elif choice == 3:
                plt.figure(figsize=(8,5))
                plt.scatter(self.data["Sales"],self.data["Profit"])
                plt.title("Sales vs Profit")
                plt.xlabel("Sales")
                plt.ylabel("Profit")
                plt.grid(True)
                plt.show()

            elif choice == 4:
                plt.figure(figsize=(7,7))
                pie = self.data.groupby("Region")["Sales"].sum()
                plt.pie(
                    pie.values,
                    labels=pie.index,
                    autopct="%1.1f%%",
                    startangle=90
                )
                plt.title("Sales Distribution by Region")
                plt.show()

            elif choice == 5:
                plt.figure(figsize=(8,5))
                plt.hist(self.data["Sales"],bins=6)
                plt.title("Sales Distribution")
                plt.xlabel("Sales")
                plt.ylabel("Frequency")
                plt.show()

            elif choice == 6:
                plt.figure(figsize=(8,5))
                yearly=self.data.groupby("Year")[["Sales","Profit"]].sum()
                plt.stackplot(
                    yearly.index,
                    yearly["Sales"],
                    yearly["Profit"],
                    labels=["Sales","Profit"]
                )
                plt.legend()
                plt.title("Sales & Profit")
                plt.show()

            elif choice == 7:
                plt.figure(figsize=(8,5))
                sns.heatmap(
                    self.data.corr(numeric_only=True),
                    annot=True,
                    cmap="Blues"
                )
                plt.title("Correlation Heatmap")
                plt.show()

            elif choice == 8:
                plt.figure(figsize=(7,5))
                sns.boxplot(data=self.data,y="Sales")
                plt.title("Sales Boxplot")
                plt.show()

            elif choice == 9:
                break

            else:
                print("Invalid Choice")

    def save_visualization(self):
        file = input("Enter file name (example graph.png): ")
        plt.savefig(file)
        print("\nVisualization Saved Successfully!")

    def menu(self):
        while True:
            print("\n=================================================")
            print("     DATA ANALYSIS & VISUALIZATION PROGRAM")
            print("=================================================")
            print("1. Load Dataset")
            print("2. Explore Data")
            print("3. Perform DataFrame Operations")
            print("4. Handle Missing Data")
            print("5. Generate Descriptive Statistics")
            print("6. Data Visualization")
            print("7. Save Visualization")
            print("8. Exit")

            try:
                choice = int(input("\nEnter your choice : "))

                if choice == 1:
                    self.load_data()

                elif choice == 2:
                    if self.data is None:
                        print("Please load dataset first.")
                    else:
                        self.explore_data()

                elif choice == 3:
                    if self.data is None:
                        print("Please load dataset first.")
                        continue

                    while True:

                        print("\n========== DATAFRAME OPERATIONS ==========")
                        print("1. Mathematical Operations")
                        print("2. Combine DataFrames")
                        print("3. Split DataFrame")
                        print("4. Search / Sort / Filter")
                        print("5. Aggregate Functions")
                        print("6. Pivot Table")
                        print("7. NumPy Operations")
                        print("8. Back")

                        op = int(input("Enter your choice : "))

                        if op == 1:
                            self.mathematical_operations()

                        elif op == 2:
                            self.combine_data()

                        elif op == 3:
                            self.split_data()

                        elif op == 4:
                            self.search_sort_filter()

                        elif op == 5:
                            self.aggregate_functions()

                        elif op == 6:
                            self.create_pivot_table()

                        elif op == 7:
                            self.numpy_operations()

                        elif op == 8:
                            break

                        else:
                            print("Invalid Choice")

                elif choice == 4:
                    if self.data is None:
                        print("Please load dataset first.")
                    else:
                        self.clean_data()

                elif choice == 5:
                    if self.data is None:
                        print("Please load dataset first.")
                    else:
                        self.statistical_analysis()

                elif choice == 6:
                    if self.data is None:
                        print("Please load dataset first.")
                    else:
                        self.visualize_data()

                elif choice == 7:
                    self.save_visualization()

                elif choice == 8:
                    print("\nThank You!")
                    print("Exiting Program...")
                    break

                else:
                    print("Invalid Choice")

            except Exception as e:
                print("Error :", e)
```

```python
obj = SalesDataAnalyzer()

obj.menu()
```
